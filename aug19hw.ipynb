## Different methods of list

## 1. append()

## Definition: Adds one item at the end of the list.

```python

a = [10, 20, 30]
a.append(40)
print(a)
```

## 2. extend()

## Definition: Adds multiple items to the end of the list.

```python
a = [10, 20, 30]
a.extend([40, 50])
print(a)
```

## 3. insert()

## Definition: Adds an item at a specific index position.

```python
a = [10, 20, 30]
a.insert(1, 15)
print(a)
```

## 4. remove()

## Definition: Removes the first occurrence of a specified value.

```python
a = [10, 20, 30, 20]
a.remove(20)
print(a)
```

## 5. pop()

## Definition: Removes and returns an item from the list. By default, it removes the last item.

```python
a = [10, 20, 30]
a.pop()
print(a)
```

## 6. clear()

## Definition: Removes all items from the list.

```python
a = [10, 20, 30]
a.clear()
print(a)
```

## 7. index()

## Definition: Returns the index position of the first occurrence of a value.

```python
a = [10, 20, 30, 40]
print(a.index(30))
```

## 8. count()

## Definition: Counts how many times a particular value occurs in the list.

```python
a = [10, 20, 30, 20, 20]
print(a.count(20))
```

## 9. sort()

## Definition: Arranges the list elements in ascending order by default.

```python
a = [40, 10, 30, 20]
a.sort()
print(a)
```

## 10. reverse()

## Definition: Reverses the order of elements in the list.

```python
a = [10, 20, 30, 40]
a.reverse()
print(a)
```

## 11. copy()

## Definition: Creates a copy of the list.

```python
a = [10, 20, 30]
b = a.copy()

print(b)
```
